```python
import cryo_ml_utils as cryo_ml
import matplotlib.pyplot as plt

import torchvision.transforms as T
```

### Load a Starfile Dataset

```python
from cryo_ml_utils.data.base.stack_dataset import ParticleStackDataset

ds = ParticleStackDataset(
    "/empiar/CryoBench/IgG-1D/images/snr0.01/snr0.01.star",
    format="channel_last",
    particle_transform=T.Compose([
        T.Normalize(0.53786290141223, 0.11803331075935841)
    ])
)
```

```python
fig, axs = plt.subplots(4, 5)
for i, ax in enumerate(axs.flat):
    img, _ = ds[i]
    ax.imshow(img.squeeze(0), cmap="gray")
    ax.set_axis_off()

fig.tight_layout()
```
